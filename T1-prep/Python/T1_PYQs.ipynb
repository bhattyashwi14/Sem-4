```python
import pandas as pd
import numpy as np
df=pd.read_csv("Car details v3.csv")
df
```

```python
# 1. How Many ritz car are there with kms driven more than 30000km.
df[(df["name"].str.contains('ritz',case=False)) & (df["km_driven"]>30000)].shape[0]
```

```python
# 2. How many Petrol cars are of the year 2017 and whose selling price > 10 lakhs.
df[(df["fuel"]=="Petrol") & (df["year"]==2017) & (df["selling_price"]>1000000)].shape[0]
```

```python
# 3. How many swi􀅌 cars are there with selling price < 4 Lakhs.
df[(df["name"].str.contains("swift",case=False)) & (df["selling_price"]<400000)].shape[0]
```

```python
# 4. How many Automa􀆟c Transmission Petrol Car are of the year 2015 whose selling price is > 10 Lakhs.
df[(df.transmission=="Automatic") & (df.fuel=="Petrol") & (df.year==2015) & (df.selling_price>1000000)].shape[0]
```

```python
# 5. List out Vehicles with Automa􀆟c Transmission and selling price < 1 Lakh.
df[(df.transmission=="Automatic") & (df.selling_price<100000)]["name"]
```

```python
# 6. How Many Petrol Vehicles are there with kms driven more than 50000kms and Year is between 2010 to 2015(both Year
# included).
df[(df.fuel=="Petrol") & (df.km_driven>50000) & (df.year>=2010) & (df.year<=2015)].shape[0]
```

```python
# 7. List out the cars whose price difference between present price and selling price is > 15 lakhs.

# no column as price difference
```

```python
# 8. How many Petrol vehicles are there whose kms driven < 5000km and selling price < 50000.
df[(df.fuel=="Petrol") & (df.km_driven<5000) & (df.selling_price<50000)].shape[0]
```

```python
# You are given a CSV file named ipl-matches.csv, which contains data of all Indian Premier League (IPL) matches played from the
# year 2008 to 2022.Read the dataset and display the basic information about the DataFrame.
# Using Python and Pandas, write code to answer the following questions:
# 1. How many matches were won by Gujarat Titans in a Super Over ?
# 2. How Many Matches won by Royal Challengers Bangalore Vs Kolkata Knight Riders.
# 3. How many Matches won by Toss Winner of the Match.
# 4. In how many matches was the result decided by the D/L method and the match was won by wickets?
# 5. In how many matches Toss won by Rajasthan Royals & Won the Match.
# 6. In how many matches V Kohli declared player of the match against Chennai Super Kings.
# 7. Which Team is the Toss Winner of the Final match Played in Dubai?
# 8. List out the Finals played by Kolkata Knight Riders in which Toss Won by Kolkata Knight Riders in the Finals and choose to
# field.
```

```python
dff=pd.read_csv("ipl-matches.csv")
dff
```

```python
# 1. How many matches were won by Gujarat Titans in a Super Over ?
dff[(dff["WinningTeam"]=="Gujarat Titans") & (dff.SuperOver=="Y")].shape[0]
```

```python
# 2. How Many Matches won by Royal Challengers Bangalore Vs Kolkata Knight Riders.
dff[((dff.Team1=="Kolkata Knight Riders") & (dff.WinningTeam=="Royal Challengers Bangalore")) | 
    ((dff.Team2=="Kolkata Knight Riders") & (dff.WinningTeam=="Royal Challengers Bangalore"))].shape[0]
```

```python
# 3. How many Matches won by Toss Winner of the Match.
dff[dff.WinningTeam==dff.TossWinner].shape[0]
```

```python
# 4. In how many matches was the result decided by the D/L method and the match was won by wickets?
dff[(dff.method=="D/L") & (dff.WonBy=="Wickets")].shape[0]
```

```python
# 5. In how many matches Toss won by Rajasthan Royals & Won the Match.
dff[(dff.TossWinner=="Rajasthan Royals") & (dff.WinningTeam=="Rajasthan Royals")].shape[0]
```

```python
# 6. In how many matches V Kohli declared player of the match against Chennai Super Kings.
dff[((dff.Team1=="Chennai Super Kings") | (dff.Team2=="Chennai Super Kings")) & (dff.Player_of_Match=="V Kohli")].shape[0]
```

```python
# 6. In how many matches V Kohli declared player of the match against Chennai Super Kings.
dff[((dff.Team1=="Chennai Super Kings") & (dff.Player_of_Match=="V Kohli")) | ((dff.Team2=="Chennai Super Kings") & (dff.Player_of_Match=="V Kohli"))].shape[0]
```

```python
# 7. Which Team is the Toss Winner of the Final match Played in Dubai?
dff[(dff.City=="Dubai") & (dff.MatchNumber=="Final")]["WinningTeam"]
```

```python
# 8. List out the Finals played by Kolkata Knight Riders in which Toss Won by Kolkata Knight Riders in the Finals and choose to
# field.
dff[(dff.MatchNumber=="Final") & (dff.TossWinner=="Kolkata Knight Riders") & (dff.TossDecision=="field")].shape[0]
```
